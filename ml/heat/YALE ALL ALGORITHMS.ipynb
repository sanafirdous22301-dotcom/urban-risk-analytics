```python
# Feature Engineering
X_yale = yale[['latitude', 'longitude', 'daytime_uhi', 'nighttime_uhi']].copy()

print("YALE FEATURE ENGINEERING")
print("------------------------")

print("Selected Features:")
print(X_yale.columns.tolist())

print("\nFeature Shape:", X_yale.shape)
print("\nFirst 5 Rows:")
print(X_yale.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Apply feature scaling
scaler_yale = StandardScaler()

X_yale_scaled = scaler_yale.fit_transform(X_yale)

print("YALE FEATURE SCALING")
print("--------------------")

print("Original Shape:", X_yale.shape)
print("Scaled Shape:", X_yale_scaled.shape)

print("\nFirst 5 Scaled Rows:")
print(X_yale_scaled[:5])
```

```python
from sklearn.cluster import KMeans

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

print("All YALE machine learning algorithms imported successfully!")
```

```python
# Configure K-Means
kmeans_yale = KMeans(
    n_clusters=7,
    random_state=42,
    n_init=10
)

print("YALE K-MEANS CONFIGURATION")
print("--------------------------")
print("Number of Clusters:", 7)
print("Random State:", 42)
print("Initialization:", 10)
```

```python
# Train K-Means
yale_clusters = kmeans_yale.fit_predict(X_yale_scaled)

# Add cluster labels to the original features
yale_clustered = X_yale.copy()
yale_clustered['Cluster'] = yale_clusters

print("YALE K-MEANS TRAINING")
print("---------------------")

print("K-Means training completed successfully!")

print("\nCluster Distribution:")
print(yale_clustered['Cluster'].value_counts().sort_index())

print("\nCluster-wise Mean Values:")
print(yale_clustered.groupby('Cluster')[[
    'daytime_uhi',
    'nighttime_uhi'
]].mean())
```

```python
# Map clusters to heat-intensity categories
cluster_mapping_yale = {
    4: 'Low',
    5: 'Low',
    2: 'Medium',
    3: 'Medium',
    6: 'Medium',
    0: 'High',
    1: 'High'
}

# Generate heat-intensity labels
yale_clustered['Heat_Intensity'] = yale_clustered['Cluster'].map(
    cluster_mapping_yale
)

print("YALE HEAT INTENSITY LABELS")
print("--------------------------")

print(yale_clustered['Heat_Intensity'].value_counts())

print("\nFirst 5 Rows:")
print(yale_clustered.head())
```

```python
# Prepare input features and target variable

X_yale = yale_clustered[
    ['latitude', 'longitude', 'daytime_uhi', 'nighttime_uhi']
]

y_yale = yale_clustered['Heat_Intensity']

print("YALE INPUT AND TARGET PREPARATION")
print("---------------------------------")

print("Input Shape:", X_yale.shape)
print("Target Shape:", y_yale.shape)

print("\nInput Features:")
print(X_yale.head())

print("\nTarget Values:")
print(y_yale.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train_yale, X_test_yale, y_train_yale, y_test_yale = train_test_split(
    X_yale,
    y_yale,
    test_size=0.20,
    random_state=42,
    stratify=y_yale
)

print("YALE TRAIN-TEST SPLIT")
print("---------------------")

print("Training Input Shape:", X_train_yale.shape)
print("Testing Input Shape:", X_test_yale.shape)

print("Training Target Shape:", y_train_yale.shape)
print("Testing Target Shape:", y_test_yale.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler_yale = StandardScaler()

# Fit and transform training data
X_train_yale_scaled = scaler_yale.fit_transform(X_train_yale)

# Transform testing data
X_test_yale_scaled = scaler_yale.transform(X_test_yale)

print("YALE SUPERVISED DATA SCALING")
print("----------------------------")

print("Training Scaled Shape:", X_train_yale_scaled.shape)
print("Testing Scaled Shape:", X_test_yale_scaled.shape)

print("\nFirst 5 Training Rows:")
print(X_train_yale_scaled[:5])
```

```python
# Initialize the six supervised models

models_yale = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'KNN': KNeighborsClassifier(),
    'SVM': SVC(),
    'Gaussian Naive Bayes': GaussianNB()
}

# Train all models
for name, model in models_yale.items():
    model.fit(X_train_yale_scaled, y_train_yale)
    print(name, "trained successfully!")

print("\nAll six YALE classification models trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Store evaluation results
results_yale = []

for name, model in models_yale.items():

    # Predict test data
    y_pred = model.predict(X_test_yale_scaled)

    # Calculate metrics
    accuracy = accuracy_score(y_test_yale, y_pred)
    precision = precision_score(y_test_yale, y_pred, average='weighted')
    recall = recall_score(y_test_yale, y_pred, average='weighted')
    f1 = f1_score(y_test_yale, y_pred, average='weighted')

    results_yale.append({
        'Algorithm': name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1
    })

# Display results
evaluation_yale = pd.DataFrame(results_yale)

print("YALE MODEL EVALUATION RESULTS")
print("-----------------------------")

print(evaluation_yale.sort_values(
    by='Accuracy',
    ascending=False
).to_string(index=False))
```

```python
from sklearn.metrics import silhouette_score

# Calculate K-Means evaluation metrics
wcss_yale = kmeans_yale.inertia_

silhouette_yale = silhouette_score(
    X_yale_scaled,
    yale_clusters
)

print("YALE K-MEANS EVALUATION")
print("-----------------------")

print("WCSS (Inertia):", wcss_yale)
print("Silhouette Score:", silhouette_yale)
```

```python
from sklearn.model_selection import GridSearchCV

print("YALE HYPERPARAMETER TUNING")
print("--------------------------")
print("GridSearchCV imported successfully!")
```

```python
# Define parameter grids for all six models

param_grids_yale = {

    'Logistic Regression': {
        'C': [0.1, 1, 10],
        'solver': ['lbfgs']
    },

    'Decision Tree': {
        'max_depth': [None, 5, 10, 15],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4]
    },

    'Random Forest': {
        'n_estimators': [100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5],
        'min_samples_leaf': [1, 2]
    },

    'KNN': {
        'n_neighbors': [3, 5, 7, 9],
        'weights': ['uniform', 'distance'],
        'metric': ['euclidean', 'manhattan']
    },

    'SVM': {
        'C': [0.1, 1, 10],
        'kernel': ['linear', 'rbf'],
        'gamma': ['scale', 'auto']
    },

    'Gaussian Naive Bayes': {
        'var_smoothing': [1e-9, 1e-8, 1e-7, 1e-6]
    }
}

print("YALE PARAMETER GRIDS")
print("--------------------")

for name, grid in param_grids_yale.items():
    print(name, ":", grid)
```

```python
# Perform hyperparameter tuning

tuned_models_yale = {}
tuning_results_yale = []

for name, model in models_yale.items():

    grid_search = GridSearchCV(
        estimator=model,
        param_grid=param_grids_yale[name],
        cv=5,
        scoring='accuracy',
        n_jobs=-1
    )

    grid_search.fit(X_train_yale_scaled, y_train_yale)

    tuned_models_yale[name] = grid_search.best_estimator_

    tuning_results_yale.append({
        'Algorithm': name,
        'Best Parameters': grid_search.best_params_,
        'Best CV Accuracy': grid_search.best_score_
    })

    print(name, "tuning completed!")

# Display tuning results
tuning_results_df_yale = pd.DataFrame(tuning_results_yale)

print("\nYALE HYPERPARAMETER TUNING RESULTS")
print("----------------------------------")

print(tuning_results_df_yale.to_string(index=False))
```

```python
from sklearn.metrics import accuracy_score

comparison_yale = []

for name, model in models_yale.items():

    # Original model accuracy
    original_pred = model.predict(X_test_yale_scaled)
    original_accuracy = accuracy_score(y_test_yale, original_pred)

    # Tuned model accuracy
    tuned_model = tuned_models_yale[name]
    tuned_pred = tuned_model.predict(X_test_yale_scaled)
    tuned_accuracy = accuracy_score(y_test_yale, tuned_pred)

    comparison_yale.append({
        'Algorithm': name,
        'Original Accuracy': original_accuracy,
        'Tuned Accuracy': tuned_accuracy
    })

comparison_df_yale = pd.DataFrame(comparison_yale)

print("YALE ORIGINAL VS TUNED MODEL ACCURACY")
print("-------------------------------------")

print(comparison_df_yale.to_string(index=False))
```
